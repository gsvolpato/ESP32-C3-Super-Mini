```python
# Save this as convert_gif.py
def convert_file_to_header(input_file, output_file):
    with open(input_file, 'rb') as f:
        content = f.read()
    
    with open(output_file, 'w') as f:
        f.write('#ifndef SPIRAL_GIF_H\n')
        f.write('#define SPIRAL_GIF_H\n\n')
        f.write('const uint8_t spiral_gif[] PROGMEM = {\n  ')
        
        # Write hex values
        for i, byte in enumerate(content):
            f.write('0x{:02x}, '.format(byte))
            if (i + 1) % 12 == 0:  # New line every 12 bytes
                f.write('\n  ')
        
        f.write('\n};\n')
        f.write('const unsigned int spiral_gif_len = {};\n\n'.format(len(content)))
        f.write('#endif\n')

# Use it
convert_file_to_header('spiral.gif', 'spiral.h')
```
